17. Hipótesis 1: ¿Los hijos aseguran el viaje? Escribe una consulta usando CASE WHEN que clasifique a la reserva como "Con Hijos" (si children > 0 o babies > 0) o "Sin Hijos". Agrupa por esta nueva categoría y calcula la tasa de cancelación (puedes sacar el promedio - AVG - de la columna is_canceled). ¿Qué segmento cancela más?

RPTA: Las familias presentan mayor estabilidad de reserva (34.92% de cancelación vs. 37.22% en huéspedes sin hijos) debido a itinerarios más rígidos. Esto permite aplicar políticas de tarifas no reembolsables a parejas o viajeros individuales.

```sql

SELECT 
    CASE 
        WHEN COALESCE(children, 0) > 0 OR COALESCE(babies, 0) > 0 THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END AS categoria_hijos,
    COUNT(*) AS total_reservas,
    ROUND(AVG(CAST(is_canceled AS DOUBLE)) * 100, 2) AS tasa_cancelacion_porcentaje
FROM cursodataacademy.proyecto_hotel.tb_reservas
GROUP BY 1
ORDER BY tasa_cancelacion_porcentaje DESC;
```

18. Hipotesis 2: El impacto de las solicitudes especiales. Agrupa por la columna total_of_special_requests y calcula la tasa de cancelación (AVG(is_canceled)) para cada cantidad de solicitudes (0, 1, 2, etc.). Ordena por la cantidad de solicitudes de forma ascendente. ¿A medida que el cliente pide más cosas, cancela más o cancela menos?

RPTA: Existe una relación inversamente proporcional directa: a mayor cantidad de peticiones especiales, la tasa de cancelación disminuye drásticamente del 47.72% (0 requerimientos) al 5.00% (5 requerimientos).

```sql

SELECT 
    total_of_special_requests,
    COUNT(*) AS total_reservas,
    ROUND(AVG(CAST(is_canceled AS DOUBLE)) * 100, 2) AS tasa_cancelacion_porcentaje
FROM cursodataacademy.proyecto_hotel.tb_reservas
GROUP BY total_of_special_requests
ORDER BY total_of_special_requests ASC;
```

19. Hipótesis 3: ¿Cambiar la reserva es señal de compromiso? Al igual que en los ejercicios anteriores, utiliza CASE WHEN para segmentar la columna booking_changes en tres grupos: "Sin Cambios" (0), "1 a 3 cambios" y "Más de 3 cambios". Calcula la tasa de cancelación para cada grupo. ¿Quiénes tienen menor riesgo de cancelar? 

RPTA: Las reservas que realizan entre 1 y 3 modificaciones registran el menor riesgo de cancelación (15.59%), en contraste con el 40.85% de tasa de anulación en aquellas que no presentan cambios.

```sql

SELECT 
    CASE 
        WHEN COALESCE(booking_changes, 0) = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        WHEN booking_changes > 3 THEN 'Más de 3 cambios'
    END AS categoria_cambios,
    COUNT(*) AS total_reservas,
    ROUND(AVG(CAST(is_canceled AS DOUBLE)) * 100, 2) AS tasa_cancelacion_porcentaje
FROM cursodataacademy.proyecto_hotel.tb_reservas
GROUP BY 1
ORDER BY tasa_cancelacion_porcentaje DESC;
```

20. Hipótesis 4: El factor precio (adr). Agrupa las reservas por tipo de hotel (hotel). Para cada tipo, muestra la cantidad total de cancelaciones y el precio promedio (adr) solo de las reservas canceladas. En base a tu consulta, ¿en qué tipo de hotel el precio promedio de cancelación es más alto?

RPTA: El Resort Hotel registra la tarifa promedio (ADR) cancelada más alta ($105.79 USD vs. $104.69 USD en City Hotel), a pesar de que City Hotel triplica el volumen de bajas (33,102 casos). El caso del Resort Hotel exige depósitos iniciales obligatorios para proteger los ingresos de mayor tarifa.

```sql
SELECT 
    hotel,
    COUNT(*) AS total_cancelaciones,
    ROUND(AVG(adr), 2) AS adr_promedio_canceladas
FROM cursodataacademy.proyecto_hotel.tb_reservas
WHERE is_canceled = 1
GROUP BY hotel
ORDER BY adr_promedio_canceladas DESC;
```
